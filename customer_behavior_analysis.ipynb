#### Creating age group column

```python
def age_group(age):
    if age<18:
        return 'Young Adult'
    if age<30:
        return 'Adult'
    if age<60:
        return 'Middle-aged'
    if age>=60:
        return 'Senior Citizen'

df['age_grp']=df['age'].apply(age_group)
```

```python
df.head(5)
```

```python
df=df.reindex(columns=['customer_id', 'age','age_grp', 'gender', 'item_purchased', 'category',
       'purchase_amount', 'location', 'size', 'color', 'season',
       'review_rating', 'subscription_status', 'shipping_type',
       'discount_applied', 'promo_code_used', 'previous_purchases',
       'payment_method', 'frequency_of_purchases'])

df.head(2)
```

```python
(df['discount_applied']==df['promo_code_used']).all()
```

This means that both discount_applied and promo_code_used column gives the same information
Hence removong promo_code_used column

```python
df=df.drop('promo_code_used',axis=1)
```
